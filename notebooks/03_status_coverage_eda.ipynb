# 03 Status Coverage EDA

상태정보 수집률과 상태 신뢰도를 분석한다.

- 상태 미수집은 사용 불가가 아니다.
- `availability_rate_among_known` 과 `unknown_status_rate` 를 분리한다.
- 전체 충전기를 분모로 단순 사용 가능률을 계산하지 않는다.


```python
from pathlib import Path
import pandas as pd

PROCESSED = Path('../apps/data-pipeline/evaluation/personal/experiments/SANDBOX_20260716_preprocess_pipeline/data/processed')
view = pd.read_csv(PROCESSED / 'charger_current_view.csv', dtype=str, low_memory=False)
status = pd.read_csv(PROCESSED / 'charger_status_current.csv', dtype=str, low_memory=False)
view['status_missing_bool'] = view['status_missing'].astype(str) == 'True'
known = view[~view['status_missing_bool']]
metrics = {
    'status_coverage_rate': 1 - view['status_missing_bool'].mean(),
    'availability_rate_among_known': (known['stat_mapped'] == 'AVAILABLE').mean(),
    'unknown_status_rate': view['status_missing_bool'].mean(),
}
metrics
```
