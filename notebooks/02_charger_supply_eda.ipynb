# 02 Charger Supply EDA

충전기 단위와 충전소 단위를 분리해 공급 현황을 분석한다.

- `charger_master`는 충전기 단위다.
- 충전소 단위 분석은 `statId` 집계 후 수행한다.
- 충전기 수가 많다는 사실과 실제 사용 가능성은 다르다.


```python
from pathlib import Path
import pandas as pd

PROCESSED = Path('../apps/data-pipeline/evaluation/personal/experiments/SANDBOX_20260716_preprocess_pipeline/data/processed')
charger = pd.read_csv(PROCESSED / 'charger_master.csv', dtype=str, low_memory=False)
charger['output_num'] = pd.to_numeric(charger['output_num'], errors='coerce')
station_supply = charger.groupby('statId').agg(
    charger_count=('chgerId', 'count'),
    station_name=('statNm', 'first'),
    lat=('lat_num', 'first'),
    lng=('lng_num', 'first'),
).reset_index()
station_supply.describe(include='all')
```
