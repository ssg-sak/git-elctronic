```python
from pathlib import Path
import numpy as np
import pandas as pd

PROCESSED = Path('../apps/data-pipeline/evaluation/personal/experiments/SANDBOX_20260716_preprocess_pipeline/data/processed')
charger = pd.read_csv(PROCESSED / 'charger_master.csv', dtype=str, low_memory=False)
poi = pd.read_csv(PROCESSED / 'poi_master.csv', dtype=str, low_memory=False)

def haversine_m(lat1, lon1, lat2, lon2):
    R = 6371000
    lat1, lon1, lat2, lon2 = map(np.radians, [lat1, lon1, lat2, lon2])
    dlat = lat2 - lat1
    dlon = lon2 - lon1
    a = np.sin(dlat/2)**2 + np.cos(lat1)*np.cos(lat2)*np.sin(dlon/2)**2
    return 2 * R * np.arcsin(np.sqrt(a))

poi['poi_type'].value_counts(dropna=False)
```
